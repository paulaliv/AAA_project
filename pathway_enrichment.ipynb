```python
def read_gmt(gmt_file):
    genes = set()

    with open(gmt_file) as f:
        for line in f:
            parts = line.strip().split("\t")
            genes.update(parts[2:])   # skip pathway name and description

    return genes
```

```python
def get_pathways(gmt_file):
    pathways = set()

    with open(gmt_file) as f:
        for line in f:
            parts = line.strip().split("\t")
            pathways.update(parts[:1])   # skip pathway name and description

    return pathways
```

```python
import gseapy as gp

tissue_pathways = gp.ssgsea(
    data=tissue_data,
    gene_sets=reactome_dir,
    outdir=None,
    sample_norm_method="rank"
)

plasma_pathways = gp.ssgsea(
    data=plasma_data,
    gene_sets=reactome_dir,
    outdir=None,
    sample_norm_method="rank"
)
```

```python
tissue_scores = tissue_pathways.res2d.T
plasma_scores = plasma_pathways.res2d.T
```

```python
print(tissue_scores)
```

```python
print(tissue_pathways.res2d.shape)
print(tissue_pathways.res2d.head())
print(tissue_pathways.res2d.columns)
```

```python
tissue_scores = tissue_pathways.res2d.pivot(
    index="Name",
    columns="Term",
    values="NES"
)
```

```python
plasma_scores = plasma_pathways.res2d.pivot(
    index="Name",
    columns="Term",
    values="NES"
)
```

```python
tissue_scores.to_csv('PR80_30/tissue_pathway_scores.csv')
plasma_scores.to_csv('PR80_30/plasma_pathway_scores.csv')
```

```python
print(tissue_scores.shape)
print(tissue_scores)
```

```python
pathway_difference = (
    tissue_scores.mean(axis=0)
    -
    plasma_scores.mean(axis=0)
)

pathway_difference.sort_values(ascending=False).head(20)
```
